2023年度の企業財務に関して分析を行う

分析結果・考察

本分析では、IT大手・中堅企業15社の財務データを用いて、売上規模と収益性・財務健全性の関係性を検証した。

売上高と営業利益率の関係を比較した結果、売上規模が大きい巨大ITベンダー（日立製作所、富士通、NTT DATA等）の営業利益率は4〜8%程度にとどまる一方、自社プロダクトや高付加価値コンサルティングを強みとする企業（オービック、野村総合研究所等）は15%〜60%超の非常に高い営業利益率を記録した。このことから、売上高の大きさと営業利益率の高さは必ずしも比例しないことが確認できた。


特にオービックは営業利益率・ROA・自己資本比率のいずれにおいても突出した数値を示しており、自社開発型の統合業務ソフトウェアを自社直接販売・サポートするビジネスモデルが極めて高い収益性と財務健全性を生み出していると考えられる。


大手SIerは豊富な総資産と売上規模を生かした大型案件の受注力に強みがあるのに対し、高収益系SIerは高い資産効率と収益率を背景とした資本効率の高さに強みがある。企業の財務的特徴を評価する際は、売上規模だけでなくビジネスモデルに応じた多角的な指標比較が必要である
